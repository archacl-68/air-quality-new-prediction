```python
df = df.dropna(subset=['Date', 'Time'])

print("Dataset shape after removing missing Date/Time:")
print(df.shape)
```

```python
df['Datetime'] = pd.to_datetime(
    df['Date'] + ' ' + df['Time'],
    dayfirst=True,
    errors='coerce'
)

print(df[['Date', 'Time', 'Datetime']].head())
```

```python
df = df.sort_values('Datetime')

df = df.reset_index(drop=True)

print(df[['Datetime', 'CO(GT)']].head())
```

```python
df['Hour'] = df['Datetime'].dt.hour
df['Day'] = df['Datetime'].dt.day
df['Month'] = df['Datetime'].dt.month
df['DayOfWeek'] = df['Datetime'].dt.dayofweek

print(df[['Datetime', 'Hour', 'Day', 'Month', 'DayOfWeek']].head())
```

```python
df['CO_Next_Hour'] = df['CO(GT)'].shift(-1)

print(
    df[['Datetime', 'CO(GT)', 'CO_Next_Hour']].head(10)
)
```

```python
df = df.dropna(subset=['CO_Next_Hour'])

print("Dataset shape:", df.shape)
```

```python
features = [
    'PT08.S1(CO)',
    'NMHC(GT)',
    'C6H6(GT)',
    'PT08.S2(NMHC)',
    'NOx(GT)',
    'PT08.S3(NOx)',
    'NO2(GT)',
    'PT08.S4(NO2)',
    'PT08.S5(O3)',
    'T',
    'RH',
    'AH',
    'Hour',
    'Day',
    'Month',
    'DayOfWeek'
]
```

```python
X = df[features]
y = df['CO_Next_Hour']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X = X.fillna(X.median())

print("Remaining missing values:")
print(X.isnull().sum().sum())
```

```python
print("Target missing values:", y.isnull().sum())
print("Target shape:", y.shape)
```

```python
split = int(len(X) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

print(model)
```

```python
model.fit(X_train, y_train)

print("Model training completed successfully!")
```

```python
y_pred = model.predict(X_test)

print("First 10 predictions:")
print(y_pred[:10])
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

print(comparison.head(10))
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error:", mae)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)
```

```python
rmse = np.sqrt(mse)

print("Root Mean Squared Error:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)
```

```python
print("MODEL PERFORMANCE")
print("-------------------------")
print("MAE  :", mae)
print("MSE  :", mse)
print("RMSE :", rmse)
print("R²   :", r2)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create an imputer
imputer = SimpleImputer(strategy='median')

# Fit on training data and transform both datasets
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Create Linear Regression model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train_imputed, y_train)

# Make predictions
linear_pred = linear_model.predict(X_test_imputed)

# Calculate performance metrics
linear_mae = mean_absolute_error(y_test, linear_pred)

linear_mse = mean_squared_error(y_test, linear_pred)

linear_rmse = np.sqrt(linear_mse)

linear_r2 = r2_score(y_test, linear_pred)

print("Linear Regression")
print("==========================")
print("MAE :", linear_mae)
print("MSE :", linear_mse)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)
```

```python
print("NaN values in X_train:", np.isnan(X_train_imputed).sum())
print("NaN values in X_test :", np.isnan(X_test_imputed).sum())
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(
    random_state=42
)

tree_model.fit(
    X_train_imputed,
    y_train
)

tree_pred = tree_model.predict(
    X_test_imputed
)

tree_mae = mean_absolute_error(
    y_test,
    tree_pred
)

tree_mse = mean_squared_error(
    y_test,
    tree_pred
)

tree_rmse = np.sqrt(tree_mse)

tree_r2 = r2_score(
    y_test,
    tree_pred
)

print("Decision Tree")
print("==========================")
print("MAE :", tree_mae)
print("MSE :", tree_mse)
print("RMSE:", tree_rmse)
print("R²  :", tree_r2)
```

```python
comparison = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Decision Tree',
        'Random Forest'
    ],
    'MAE': [
        linear_mae,
        tree_mae,
        mae
    ],
    'RMSE': [
        linear_rmse,
        tree_rmse,
        rmse
    ],
    'R2 Score': [
        linear_r2,
        tree_r2,
        r2
    ]
})

print(comparison)
```
